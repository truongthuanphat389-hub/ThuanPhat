```python
# Câu 1
# ================================
# BÀI TOÁN TÔ MÀU BẢN ĐỒ
# Thuật toán tham lam (Greedy Coloring)
# ================================

# Danh sách các vùng kề nhau
graph = {
    'A': ['D', 'F', 'I'],
    'B': ['H', 'J'],
    'C': ['F', 'G', 'I', 'J', 'K'],
    'D': ['A', 'E', 'I'],
    'E': ['D', 'I', 'K', 'L'],
    'F': ['A', 'C', 'I'],
    'G': ['C', 'J'],
    'H': ['B', 'J', 'K', 'L'],
    'I': ['A', 'C', 'D', 'E', 'F', 'K'],
    'J': ['B', 'C', 'G', 'H', 'K'],
    'K': ['C', 'E', 'H', 'I', 'J', 'L'],
    'L': ['E', 'H', 'K']
}

# Danh sách màu cố định
colors = ["Blue", "Red", "Yellow", "Green"]

# ==========================================
# a) TÍNH BẬC CỦA TỪNG VÙNG
# ==========================================
degree = {}
for region in graph:
    degree[region] = len(graph[region])
print("===== BẬC CỦA TỪNG VÙNG =====")
for region in degree:
    print(region, ":", degree[region])
regions = list(graph.keys())
for i in range(len(regions)):
    max_index = i
    for j in range(i + 1, len(regions)):
        if degree[regions[j]] > degree[regions[max_index]]:
            max_index = j
    # Hoán vị
    regions[i], regions[max_index] = regions[max_index], regions[i]
print("\n===== THỨ TỰ TÔ MÀU =====")
print(regions)
# ==========================================
# b) + c) TÔ MÀU BẢN ĐỒ
# ==========================================
result = {}
print("\n===== QUÁ TRÌNH TÔ MÀU =====")
for region in regions:
    # Danh sách màu bị cấm
    forbidden_colors = []
    # Kiểm tra các vùng kề đã tô
    for neighbor in graph[region]:
        if neighbor in result:
            forbidden_colors.append(result[neighbor])
    # Chọn màu đầu tiên hợp lệ
    for color in colors:
        if color not in forbidden_colors:
            result[region] = color
            break
    # In truy vết
    print("\n--------------------------------")
    print("Tô vùng:", region)
    print("Màu được gán:", result[region])
    print("Các vùng kề bị loại màu", result[region], ":")
    for neighbor in graph[region]:
        print(neighbor, end=" ")
    print()
# ==========================================
# c) KẾT QUẢ CUỐI CÙNG
# ==========================================
print("\n===== KẾT QUẢ TÔ MÀU CUỐI CÙNG =====")
for region in sorted(result):
    print(region, "->", result[region])
# Đếm số màu đã dùng
used_colors = set(result.values())
print("\nSố màu đã sử dụng:", len(used_colors))
print("Danh sách màu đã dùng:")
for color in used_colors:
    print("-", color)
# ==========================================
# d) NHẬN XÉT
# ==========================================
print("\n===== NHẬN XÉT =====")
print("Thuật toán tham lam đã sử dụng", len(used_colors), "màu.")
print("Tuy nhiên đây chưa chắc là số màu tối thiểu.")
print("Bản đồ này có thể tô bằng 3 màu nếu chọn")
print("thứ tự tô màu khác hợp lý hơn.")
print("Thuật toán tham lam chỉ chọn tối ưu cục bộ")
print("nên đôi khi không cho kết quả tối ưu toàn cục.")
```

```python
# CÂU 2
import math
from collections import Counter
# =========================================================
# HÀM TÍNH KHOẢNG CÁCH EUCLID
def euclid(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)
# PHẦN A - K-MEANS
print("========== PHẦN A : K-MEANS ==========\n")
# ---------------------------------------------------------
# DỮ LIỆU
# ---------------------------------------------------------
points = {
    "P1": (1, 2),
    "P2": (2, -1),
    "P3": (3, 0),
    "P4": (7, 1),
    "P5": (6, -2),
    "P6": (8, 0)
}
# Tâm cụm ban đầu
C1 = (0, 0)
C2 = (6, 0)
print("Tâm ban đầu:")
print("C1 =", C1)
print("C2 =", C2)
# ---------------------------------------------------------
# a) GÁN NHÃN CHO CÁC ĐIỂM
# ---------------------------------------------------------
print("\n========== a) GÁN NHÃN ==========\n")
cluster1 = []
cluster2 = []
for name, point in points.items():
    d1 = euclid(point, C1)
    d2 = euclid(point, C2)
    print(name, point)
    print("Khoảng cách tới C1 =", round(d1, 2))
    print("Khoảng cách tới C2 =", round(d2, 2))
    # np.argmin sẽ chọn phần tử đầu tiên nếu bằng nhau
    if d1 <= d2:
        cluster1.append(point)
        print("=> Gán vào cụm C1\n")
    else:
        cluster2.append(point)
        print("=> Gán vào cụm C2\n")
# ---------------------------------------------------------
# GIẢI THÍCH RIÊNG P3
# ---------------------------------------------------------
print("Giải thích điểm P3:")
print("P3 = (3,0)")
print("Khoảng cách tới C1 = 3")
print("Khoảng cách tới C2 = 3")
print("Vì hai khoảng cách bằng nhau nên np.argmin")
print("sẽ chọn tâm xuất hiện trước.")
print("=> P3 được gán vào cụm C1\n")
# ---------------------------------------------------------
# b) CẬP NHẬT TÂM CỤM
# ---------------------------------------------------------
print("========== b) CẬP NHẬT TÂM ==========\n")
# Tính tâm mới cho cụm 1
x1 = sum(p[0] for p in cluster1) / len(cluster1)
y1 = sum(p[1] for p in cluster1) / len(cluster1)
new_C1 = (x1, y1)
# Tính tâm mới cho cụm 2
x2 = sum(p[0] for p in cluster2) / len(cluster2)
y2 = sum(p[1] for p in cluster2) / len(cluster2)
new_C2 = (x2, y2)
print("Cụm 1:", cluster1)
print("Tâm mới C1 =", new_C1)
print()
print("Cụm 2:", cluster2)
print("Tâm mới C2 =", new_C2)
# ---------------------------------------------------------
# c) KIỂM TRA HỘI TỤ
# ---------------------------------------------------------
print("\n========== c) KIỂM TRA HỘI TỤ ==========\n")
if C1 == new_C1 and C2 == new_C2:
    print("Thuật toán đã hội tụ.")
else:
    print("Thuật toán CHƯA hội tụ.")
print()
print("Vì tâm mới khác tâm cũ:")
print("C1 cũ =", C1, "-> C1 mới =", new_C1)
print("C2 cũ =", C2, "-> C2 mới =", new_C2)
# PHẦN B - KNN
# =========================================================
# =========================================================
print("\n\n========== PHẦN B : K-NN ==========\n")
train_data = [
    ("T1", (2, 0), "Lớp 0"),
    ("T2", (2, 2), "Lớp 1"),
    ("T3", (3, 0), "Lớp 0"),
    ("T4", (2, 3), "Lớp 1"),
    ("T5", (4, 0), "Lớp 1"),
    ("T6", (4, 2), "Lớp 0")
]
# Điểm cần phân loại
q = (0, 0)
print("Điểm cần phân loại q =", q)
# ---------------------------------------------------------
# a) TÍNH KHOẢNG CÁCH
# ---------------------------------------------------------
print("\n========== a) TÍNH KHOẢNG CÁCH ==========\n")
distances = []
for name, point, label in train_data:
    d = euclid(q, point)
    distances.append((d, name, label))
    print(name, point, label)
    print("Khoảng cách =", round(d, 2))
    print()
# Sắp xếp tăng dần
distances.sort()
print("========== SAU KHI SẮP XẾP ==========\n")
for item in distances:
    print(item)
# ---------------------------------------------------------
# b) K = 3
# ---------------------------------------------------------
print("\n========== b) K = 3 ==========\n")
k = 3
neighbors = distances[:k]
print("3 láng giềng gần nhất:")
for n in neighbors:
    print(n)
labels = [n[2] for n in neighbors]
print("\nCác nhãn:", labels)
result = Counter(labels).most_common(1)[0][0]
print("=> q thuộc:", result)
# ---------------------------------------------------------
# c) K = 4
# ---------------------------------------------------------
print("\n========== c) K = 4 ==========\n")
k = 4
neighbors = distances[:k]
print("4 láng giềng gần nhất:")
for n in neighbors:
    print(n)
labels = [n[2] for n in neighbors]
print("\nCác nhãn:", labels)
# Đếm số lần xuất hiện
count = Counter(labels)
print("Số lần xuất hiện:")
print(count)
# Mô phỏng:
# max(classes, key=classes.get)
result = max(count, key=count.get)
print("\nKết quả chọn:", result)
print()
print("Giải thích:")
print("- Nếu hòa phiếu, max() sẽ chọn nhãn xuất hiện")
print("  trước trong dictionary.")
print("- Vì vậy kết quả phụ thuộc thứ tự thêm nhãn.")
# ---------------------------------------------------------
# K = 5
# ---------------------------------------------------------
print("\n========== K = 5 ==========\n")
k = 5
neighbors = distances[:k]
labels = [n[2] for n in neighbors]
count = Counter(labels)
print("5 láng giềng gần nhất:")
for n in neighbors:
    print(n)
print("\nCác nhãn:", labels)
print("Số lần xuất hiện:")
print(count)
result = max(count, key=count.get)
print("\n=> q thuộc:", result)
print()
print("Giải thích:")
print("K = 5 không còn hòa phiếu.")
print("Lớp xuất hiện nhiều hơn sẽ được chọn.")
```
